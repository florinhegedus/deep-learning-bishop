```python
import torch
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])

train_dataset = datasets.OxfordIIITPet(
    root="./data",
    split="trainval",
    target_types="category",
    transform=transform,
    download=True,
)

test_dataset = datasets.OxfordIIITPet(
    root="./data",
    split="test",
    target_types="category",
    transform=transform,
    download=True,
)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)
```

```python
images, labels = next(iter(train_loader))
```

```python
import matplotlib.pyplot as plt

print(set(labels))
print(images[0].shape)
image = images[0].permute(1,2,0)
plt.imshow(image)
len(test_dataset.classes)
```

```python
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

class Net(nn.Module):
    def __init__(self):

        # input shape : 3x224x224
        super().__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=6, kernel_size=3, stride=2, padding=1) # 6x112x112
        self.conv2 = nn.Conv2d(in_channels=6, out_channels=12, kernel_size=3, stride=2, padding=1) # 12x56x56
        self.conv3 = nn.Conv2d(in_channels=12, out_channels=24, kernel_size=3, stride=2, padding=1) # 24x28x28
        self.conv4 = nn.Conv2d(in_channels=24, out_channels=48, kernel_size=3, stride=2, padding=1) # 48x14x14
        
        self.fc1 = nn.Linear(48*14*14, 120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 37)

    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = F.relu(self.conv2(x))
        x = F.relu(self.conv3(x))
        x = F.relu(self.conv4(x))
        x = torch.flatten(x, 1) # flatten all dimensions except batch
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

```python
from torchinfo import summary
model = Net()
summary(model)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
model.train()
epochs = 5
losses = []

for epoch in epochs:
    for imgs, labels in train_loader:
        pred = model(imgs)
        print(f'pred shape: {pred.shape}')
        print(f'labels shape: {labels.shape}')
        loss = criterion(pred, labels)
        
        losses.append(loss)
        loss.backward()
        optimizer.step()
    print(f'loss after epoch {epoch} is {loss}')
    losses.append(loss)
```
